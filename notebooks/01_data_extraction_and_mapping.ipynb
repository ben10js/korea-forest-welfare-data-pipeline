## Prepare `df sheet3` for counting

### Subtask:
Identify the 'types' in `df_sheet3` that correspond to the column names in `df_sheet2`. This might involve inspecting the '기관구분코드명' column in `df_sheet3`.


**Reasoning**:
Inspect the column names of df_sheet2 and the unique values in the '기관구분코드명' column of df_sheet3 to identify the corresponding types.



```python
# Identify type columns in df_sheet2 (excluding sgg_name)
type_columns_sheet2 = df_sheet2.columns[1:8].tolist()

# Get unique values from '기관구분코드명' in df_sheet3
unique_types_sheet3 = df_sheet3['기관구분코드명'].unique().tolist()

print("Type columns in df_sheet2:", type_columns_sheet2)
print("Unique values in '기관구분코드명' (df_sheet3):", unique_types_sheet3)
```

**Reasoning**:
Compare the identified type columns from df_sheet2 with the unique values from df_sheet3 to confirm the correspondence and identify any potential discrepancies.



```python
# Compare the two lists
matching_types = [col for col in type_columns_sheet2 if col in unique_types_sheet3]
non_matching_sheet2 = [col for col in type_columns_sheet2 if col not in unique_types_sheet3]
non_matching_sheet3 = [val for val in unique_types_sheet3 if val not in type_columns_sheet2]

print("Matching types:", matching_types)
print("Types in df_sheet2 not found in df_sheet3:", non_matching_sheet2)
print("Types in df_sheet3 not found in df_sheet2:", non_matching_sheet3)
```

## Count types in `df sheet3`

### Subtask:
For each row (which corresponds to a location) in `df_sheet3`, count the occurrences of each 'type' that matches a column name in `df_sheet2`.


**Reasoning**:
Implement steps 1-6 from the instructions to count the occurrences of each type for each location in df_sheet3.



```python
# Step 1: Create a mapping dictionary
type_mapping = {
    '국립휴양': '국립자연휴양림',
    '공립휴양': '공립자연휴양림',
    '사립휴양': '사립자연휴양림',
    '국립숲길': '국립숲길',
    '공립숲길': '공립숲길',
    '국립레포츠': '국립레포츠',
    '공립레포츠': '공립레포츠'
}

# Step 2: Initialize a dictionary to store counts by location and type
location_type_counts = {}

# Step 3-6: Iterate through each row of df_sheet3, identify type and location, and count
for index, row in df_sheet3.iterrows():
    기관_type_sheet3 = row['기관구분코드명']
    도로명주소 = row['도로명주소']

    # Extract location information (e.g., city/district) from '도로명주소'
    # This is a simplification; a more robust approach might involve a geocoding library
    location = ' '.join(도로명주소.split(' ')[:2]) # Using the first two parts of the address as location

    # Find the corresponding type column name from df_sheet2 using the mapping
    type_sheet2 = None
    for key, value in type_mapping.items():
        if value == 기관_type_sheet3:
            type_sheet2 = key
            break

    if type_sheet2:
        # Initialize location in the counts dictionary if not present
        if location not in location_type_counts:
            location_type_counts[location] = {col: 0 for col in type_mapping.keys()}

        # Increment the count for the identified type and location
        location_type_counts[location][type_sheet2] += 1

# Display the resulting counts for each location and type
for location, counts in location_type_counts.items():
    print(f"Location: {location}")
    for type_name, count in counts.items():
        print(f"  {type_name}: {count}")
```

## Create a new dataframe for counts

### Subtask:
Create a new DataFrame that has the same index (row names) and columns (type names) as `df_sheet2`.


**Reasoning**:
Create an empty DataFrame with the row index and column names from df_sheet2, excluding irrelevant columns.



```python
# Get the index (row names) from df_sheet2
df_sheet2_index = df_sheet2['sgg_name']

# Get the column names that represent the types from df_sheet2
# These are the columns from the second column onwards, excluding any summary or irrelevant columns.
# Based on the head of df_sheet2, columns 1 to 7 (0-indexed) seem to be the type columns.
type_columns = df_sheet2.columns[1:8].tolist()

# Create an empty DataFrame with the identified index and column names
merged_df = pd.DataFrame(0, index=df_sheet2_index, columns=type_columns)

# Display the head of the newly created empty DataFrame
display(merged_df.head())
```

## Populate the new dataframe with counts

### Subtask:
Map the counts obtained in step 2 (stored in `location_type_counts`) to the corresponding cells in the new DataFrame created in step 3 (`merged_df`).


**Reasoning**:
Iterate through the calculated counts and populate the `merged_df` by matching the locations and type columns.



```python
# Iterate through the location_type_counts dictionary
for location, counts in location_type_counts.items():
    # Format the location to match the sgg_name format in merged_df
    # This is a simplification and might need adjustment based on actual data variations
    # For now, we'll assume a simple match based on the derived location string
    # A more robust approach would involve mapping or fuzzy matching

    # --- Added specific mapping for '세종특별자치시 금남면' ---
    matched_sgg_name = None
    if location == '세종특별자치시 금남면':
        matched_sgg_name = '세종' # Map to '세종' in df_sheet2 index
    else:
        # Split the location string and get the last part (e.g., '가평군' from '경기 가평군')
        location_parts = location.split()
        if len(location_parts) > 1:
            sgg_part = location_parts[-1]

            # Find matching index entries in merged_df
            matching_indices = merged_df.index[merged_df.index.str.contains(sgg_part, na=False)].tolist()

            if matching_indices:
                # Assuming the first match is the correct one for simplicity
                # In a real scenario, you might need a more sophisticated matching logic
                matched_sgg_name = matching_indices[0]
        else:
             print(f"Warning: Location '{location}' from df_sheet3 could not be parsed for matching.")
    # --- End of added specific mapping ---


    if matched_sgg_name and matched_sgg_name in merged_df.index:
        # If a matching sgg_name is found, update the counts in merged_df
        for type_name, count in counts.items():
            # Ensure the type_name is a valid column in merged_df
            if type_name in merged_df.columns:
                merged_df.loc[matched_sgg_name, type_name] += count # Use += to sum counts for locations mapping to the same sgg_name
    elif matched_sgg_name is None:
         # Handle locations from df_sheet3 that don't have a direct match in df_sheet2's index
         # and were not covered by specific mappings
         print(f"Warning: Location '{location}' from df_sheet3 not found in df_sheet2 index and no specific mapping applied.")


# Display the head of the updated merged_df to verify the counts are populated
display(merged_df.head())

# Display some rows with non-zero counts to further verify
display(merged_df[merged_df.sum(axis=1) > 0].head())
```
