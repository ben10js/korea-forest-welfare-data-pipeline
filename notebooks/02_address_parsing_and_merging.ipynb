```python
item_column = df_forest['구 분']
completion_year_column = df_forest['준공연도']

budget_columns = df_forest.iloc[:, 7:37]

print("Head of selected '항목' column:")
display(item_column.head())

print("\nHead of selected '준공연도' column:")
display(completion_year_column.head())

print("\nHead of selected '운용 예산(백만원)' columns:")
display(budget_columns.head())
```

```python
df_selected = pd.concat([item_column, completion_year_column, budget_columns], axis=1)

print("Head of the new DataFrame with selected columns:")
display(df_selected.head())

print("\nTail of the new DataFrame with selected columns:")
display(df_selected.tail())
```

```python
df_cleaned = df_selected.iloc[2:].copy()

df_cleaned['준공연도'] = pd.to_numeric(df_cleaned['준공연도'].astype(str).str.replace('년', '').str.strip(), errors='coerce')

budget_col_names = df_cleaned.columns[2:]

budget_years_row = df_selected.iloc[0, 2:]
budget_years = pd.to_numeric(budget_years_row.astype(str).str.replace('.0', '', regex=False).str.strip(), errors='coerce')

col_to_year = {col_name: year for col_name, year in zip(budget_col_names, budget_years)}

transformed_data = []

for index, row in df_cleaned.iterrows():
    item = row['구 분']
    completion_year = row['준공연도']

    if pd.isna(completion_year):
        relevant_budget_values = []
    else:
        # 요청할 예산이 존재하는 곳들 고르기
        relevant_budget_cols_for_row = [col for col in budget_col_names if col_to_year.get(col, float('inf')) >= completion_year and col_to_year.get(col, float('-inf')) <= 2024]
        relevant_budget_values = row[relevant_budget_cols_for_row].tolist()

    transformed_data.append([item, completion_year, relevant_budget_values])


new_column_names = ['구 분', '준공연도', '운용 예산(백만원)']


df_transformed = pd.DataFrame(transformed_data, columns=new_column_names)


print("Head of the transformed DataFrame:")
display(df_transformed.head())

print("\nTail of the transformed DataFrame:")
display(df_transformed.tail())

print(f"\nShape of the transformed DataFrame: {df_transformed.shape}")
```

```python
import io
import pandas as pd
full_adress = B[4:,1]

full_adress.shape
```

```python
# 광역시도a, 시군구명b, 읍면동면c, 상세주소(전체)
# Initialize lists to store the split parts
a_list = [] # 광역시도
b_list = [] # 시군구명
c_list = [] # 읍면동명
d_list = [] # 상세주소 (remaining parts)

for address_str in full_adress:
    # Split the address string by space, limiting to 3 splits
    # This ensures that any remaining parts of the address stay together in the last element
    parts = address_str.split(' ', 3)

    # Append parts to respective lists, handling cases where there might be fewer than 4 parts
    a_list.append(parts[0] if len(parts) > 0 else '')
    b_list.append(parts[1] if len(parts) > 1 else '')
    c_list.append(parts[2] if len(parts) > 2 else '')
    d_list.append(parts[3] if len(parts) > 3 else '')

# Convert lists to numpy arrays
a = np.array(a_list)
b = np.array(b_list)
c = np.array(c_list)
d = np.array(d_list)

print("광역시도 (a):")
print(a)
print("\n시군구명 (b):")
print(b)
print("\n읍면동명 (c):")
print(c)
print("\n상세주소 (d):")
print(d)
```

```python
# Reshape a, b, and c into column vectors
a_col = a.reshape(-1, 1)
b_col = b.reshape(-1, 1)
c_col = c.reshape(-1, 1)
full_adress_col = full_adress.reshape(-1,1)

# Concatenate them horizontally (column-wise)
concatenated_abc_cols = np.concatenate((a_col, b_col, c_col, full_adress_col), axis=1)

print("Concatenated array (a, b, c) as columns (first 5 rows):")
print(concatenated_abc_cols[:5])
print("\nShape of the concatenated array:")
print(concatenated_abc_cols.shape)
```
