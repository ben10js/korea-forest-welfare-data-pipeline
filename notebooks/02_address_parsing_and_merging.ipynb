## Create the new table

### Subtask:
Construct a new DataFrame using the extracted columns, with '항목' and '준공연도' as direct columns and the budget columns transformed to represent '각 준공연도' to 2024.


**Reasoning**:
Drop the first two header rows, convert '준공연도' to numeric, identify budget columns, iterate through rows to select budget data from the completion year to 2024, and construct the final DataFrame with appropriate column names.



```python
# 1. Drop the first two rows of df_selected
df_cleaned = df_selected.iloc[2:].copy()

# 2. Convert the '준공연도' column to numeric
# Handle potential errors by coercing invalid parsing to NaN and removing '년'
df_cleaned['준공연도'] = pd.to_numeric(df_cleaned['준공연도'].astype(str).str.replace('년', '').str.strip(), errors='coerce')

# 3. Identify the columns that contain budget data (from the third column onwards)
# The first two columns are '구 분' and '준공연도'
budget_col_names = df_cleaned.columns[2:]

# Define the range of years for budget data based on the first row of the original df_selected
# The years are in the first row of the budget columns
budget_years_row = df_selected.iloc[0, 2:]
budget_years = pd.to_numeric(budget_years_row.astype(str).str.replace('.0', '', regex=False).str.strip(), errors='coerce')


# Map column names to years based on the extracted budget_years
col_to_year = {col_name: year for col_name, year in zip(budget_col_names, budget_years)}

# 4. & 5. Iterate through the rows and select relevant budget data
transformed_data = []

for index, row in df_cleaned.iterrows():
    item = row['구 분']
    completion_year = row['준공연도']

    # Ensure completion_year is a valid number
    if pd.isna(completion_year):
        # If completion year is missing or invalid, we cannot determine the budget range
        # Append None for budget values or handle as appropriate for the task
        # For this task, we will append empty list for budget values
        relevant_budget_values = []
    else:
        # Select budget columns for years >= completion_year and <= 2024
        relevant_budget_cols_for_row = [col for col in budget_col_names if col_to_year.get(col, float('inf')) >= completion_year and col_to_year.get(col, float('-inf')) <= 2024]
        relevant_budget_values = row[relevant_budget_cols_for_row].tolist()

    # 6. Append the transformed data for the row
    # Combine the item, completion year and the list of budget values into a single list
    transformed_data.append([item, completion_year, relevant_budget_values])

# Determine the maximum number of budget columns needed across all rows
# This is not needed anymore as the budget values are in a list within a single column

# Define the column names for the new DataFrame
new_column_names = ['구 분', '준공연도', '운용 예산(백만원)']


# 7. Create a new DataFrame from the transformed data list
df_transformed = pd.DataFrame(transformed_data, columns=new_column_names)


# 8. Column names are already defined in step 7

print("Head of the transformed DataFrame:")
display(df_transformed.head())

print("\nTail of the transformed DataFrame:")
display(df_transformed.tail())

print(f"\nShape of the transformed DataFrame: {df_transformed.shape}")
```

```python
# Update the '운용 예산(백만원)' column
# Convert '준공연도' to string, remove '.0' and handle potential NaN values before combining with "~2024"
df_transformed['운용 예산(백만원)'] = df_transformed['준공연도'].astype(str).str.replace('.0', '', regex=False).replace('nan', '', regex=False) + '~2024'

print("DataFrame after updating '운용 예산(백만원)' column:")
display(df_transformed.head())
display(df_transformed.tail())
```

```python
# 광역시도a, 시군구명b, 읍면동면c, 상세주소(전체)
# Initialize lists to store the split parts
a_list = [] # 광역시도
b_list = [] # 시군구명
c_list = [] # 읍면동명
d_list = [] # 상세주소 (remaining parts)

for address_str in full_adress:
    # Split the address string by space, limiting to 3 splits
    # This ensures that any remaining parts of the address stay together in the last element
    parts = address_str.split(' ', 3)

    # Append parts to respective lists, handling cases where there might be fewer than 4 parts
    a_list.append(parts[0] if len(parts) > 0 else '')
    b_list.append(parts[1] if len(parts) > 1 else '')
    c_list.append(parts[2] if len(parts) > 2 else '')
    d_list.append(parts[3] if len(parts) > 3 else '')

# Convert lists to numpy arrays
a = np.array(a_list)
b = np.array(b_list)
c = np.array(c_list)
d = np.array(d_list)

print("광역시도 (a):")
print(a)
print("\n시군구명 (b):")
print(b)
print("\n읍면동명 (c):")
print(c)
print("\n상세주소 (d):")
print(d)
```
